```python
import pandas as pd
import requests
import time

df = pd.DataFrame()
max_pages = 502  # Default cap

for i in range(1, 503):
    url = 'https://api.themoviedb.org/3/movie/top_rated?api_key=c35572b7f8f9f845d08386292a881b3c&language=en-US&page={}'.format(i)
    responds = requests.get(url)
    
    if responds.status_code == 200:
        data = responds.json()
        if 'results' in data:
            if i == 1:
                max_pages = min(data['total_pages'], 502)  # Set max based on API
            df_lol = pd.DataFrame(data['results'])[['id', 'original_language', 'title', 'popularity', 'release_date', 'vote_average', 'vote_count']]
            df = pd.concat([df, df_lol], ignore_index=True)
        else:
            print(f"Page {i}: No results. Stopping. Response: {data}")
            break
    else:
        print(f"Page {i}: Failed with status {responds.status_code}. Response: {responds.text}")
        break
    
    if i >= max_pages:
        print(f"Reached max pages ({max_pages})")
        break
    
    time.sleep(0.25)  # Avoid rate limiting

print(f"Collected {len(df)} movies")
```
